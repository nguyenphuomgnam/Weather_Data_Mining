```python
import os
import sys
import pandas as pd

if os.getcwd().endswith('notebooks'):
    os.chdir('..')
sys.path.append(os.getcwd())

from src.association import WeatherAssociation

# 1. Đọc lại dữ liệu đã làm sạch từ Bước 1
import yaml
with open("configs/params.yaml", "r", encoding="utf-8") as f:
    config = yaml.safe_load(f)
df_clean = pd.read_csv(config['data']['processed_path'])

# 2. Khởi tạo module Rời rạc hoá và Khai phá luật
assoc_miner = WeatherAssociation()

# 3. Rời rạc hoá dữ liệu
df_discrete = assoc_miner.discretize_data(df_clean)

# 4. Tìm luật cho Mùa Hè
rules_summer = assoc_miner.run_fpgrowth(df_discrete, season="Summer")

# 5. Xem Top 5 luật mạnh nhất
rules_summer[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head()
```

```python
seasons = ['Spring', 'Summer', 'Autumn', 'Winter']

for s in seasons:
    print(f"\n{'='*40}")
    print(f"🌤️ TOP 3 LUẬT MẠNH NHẤT MÙA: {s.upper()}")
    print(f"{'='*40}")
    
    rules = assoc_miner.run_fpgrowth(df_discrete, season=s)
    
    if not rules.empty:
        # Lấy top 3 luật có Lift cao nhất
        display(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head(3))
    else:
        print("Không tìm thấy luật nào thỏa mãn ngưỡng (Support/Confidence quá cao).")
```
